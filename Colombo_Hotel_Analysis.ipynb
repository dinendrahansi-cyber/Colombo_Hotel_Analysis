```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create a dictionary with hotel data
data = {
    'Hotel_Name': ['Cinnamon Grand', 'Kingsbury', 'Galadari', 'Shangri-La', 'Hilton'],
    'Monthly_Revenue_M': [25, 18, 12, 30, 22],
    'Average_Rating': [4.5, 4.2, 3.9, 4.8, 4.6]
}

# 2. Convert the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# 3. Plot a bar chart for Monthly Revenue
plt.figure(figsize=(8, 4))
plt.bar(df['Hotel_Name'], df['Monthly_Revenue_M'], color='teal')
plt.title('Monthly Revenue of Top Hotels in Colombo (LKR Millions)')
plt.xlabel('Hotel Name')
plt.ylabel('Revenue (LKR Millions)')
plt.show()
```

```python
# Calculate the average monthly revenue
average_revenue = df['Monthly_Revenue_M'].mean()

print("The Average Monthly Revenue of Colombo Hotels is:")
print(average_revenue, "LKR Millions")
```
